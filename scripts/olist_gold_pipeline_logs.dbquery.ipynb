```sql
SELECT
    pipeline_run_id,
    COUNT(*) AS total_stages,
    SUM(CASE WHEN status = 'SUCCESS' THEN 1 ELSE 0 END) AS successful_stages,
    SUM(CASE WHEN status = 'FAILED' THEN 1 ELSE 0 END) AS failed_stages,
    ROUND(SUM(duration_seconds), 2) AS total_duration_seconds,
    MIN(start_time_utc) AS pipeline_started_at,
    MAX(end_time_utc) AS pipeline_completed_at
FROM olist_data_quality.gold.pipeline_run_log
GROUP BY pipeline_run_id
ORDER BY pipeline_started_at DESC;
--monitor data quality
SELECT
    run_id,
    check_name,
    table_name,
    failed_records,
    status,
    run_timestamp_utc
FROM olist_data_quality.gold.data_quality_audit
ORDER BY run_timestamp_utc DESC;
```

```sql
SELECT
    order_id,
    payment_type,
    payment_value
FROM olist_data_quality.bronze.olist_order_payments_dataset
WHERE payment_value IS NOT NULL
  AND (
      TRY_CAST(payment_value AS DOUBLE) IS NULL
      OR TRY_CAST(payment_value AS DOUBLE) <= 0
  );
---------------------------------------------------------------------------------
SELECT
    payment_value,
    COUNT(*) AS record_count
FROM olist_data_quality.bronze.olist_order_payments_dataset
GROUP BY payment_value
ORDER BY TRY_CAST(payment_value AS DOUBLE);
```
